```python
import fsspec
import pyarrow.parquet as pq

from tqdm.auto import tqdm

first = True

with fsspec.open(url, mode='rb') as file:
    pf = pq.ParquetFile(file)

    with engine.begin() as conn:
        for batch in tqdm(pf.iter_batches(batch_size=10000)):
            batch_df = batch.to_pandas()

            if first:
                batch_df.to_sql(name='green_tripdata', con=engine, if_exists='replace')
                first = False
            else:
                batch_df.to_sql(name='green_tripdata', con=engine, if_exists='append')
```
